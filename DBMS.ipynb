# **What is RDBMS**

---

RDBMS stands for relational Database Management System.


***Relational*** : Data is organized into tables (which are related to each other).

***Database :*** A collection of structure data.

***Management system :*** Software thar allows you to create , read , update , and delete data(using SQL).

An **RDBMS** is software that helps you store data in tables, relate them to each other, and perform operations on them using SQL.

* Example - PostgreSQL, MySQL, SQLite, Oracle Database, SQL Server.

# **PostgreSQL**:

---


PostgreSQL is an open-source relational database management system. It uses SQL to store, manage, and retrieve structured data. It supports features such as transactions, relationships, constraints, complex queries, and
advanced data types. It is commonly used for web applications, Banking System, Data Analytics, enterprise systems, and data-intensive applications.

* Why PostgreSQL:
   * Open Source
   * Most feature-rich & powerful
   * Supports complex queries, joins, CTEs, window functions
   * ACID compliant
   * Highly extensible (you can add your own data types, operators, etc.)
   * Support for NoSQL features also (JSONB, hstore).
   * Used for enterprise-grade applications.

#**CURD Operations:**

---



CRUD stands for Create, Read, Update, and Delete.

1. Create (C)

   Definition: Create is used to insert new data into a database.

   Example: Adding a new student's information to the student table.

2. Read (R)

   Definition: Read is used to retrieve or view data from a database.

   Example: Displaying all student records.

3. Update (U)

   Definition: Update is used to modify existing data in a database.

   Example: Changing a student's age.


4. Delete (D)

   Definition: Delete is used to remove existing records from a database.

   Example: Deleting a student's record.

```python
%%sql

# for create
INSERT INTO students (id, name, age)
VALUES (1, 'Anupam', 21);

# read
SELECT * FROM students;

# update
UPDATE students
SET age = 22
WHERE id = 1;

# delete
DELETE FROM students
WHERE id = 1;
```

#**Constraints**

---
SQL Constraints are rules applied to table columns to maintain the accuracy, validity, and integrity of data in a database.




| Constraint | Meaning | Example |
|---|---|---|
| PRIMARY KEY | Uniquely identifies each row | `id SERIAL PRIMARY KEY` |
| NOT NULL | Column must have a value | `name TEXT NOT NULL` |
| UNIQUE | No duplicate values allowed | `email TEXT UNIQUE` |
| DEFAULT | Provides default value if none is given | `created_at TIMESTAMP DEFAULT now()` |
| CHECK | Validates values based on a condition | `age INT CHECK (age > 18)` |
| FOREIGN KEY | Links one table to another | `user_id INT REFERENCES users(id)` |
